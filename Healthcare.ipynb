```python
import random
import matplotlib.pyplot as plt

class CognitiveComputingHealthcare:
    def __init__(self):
        self.diseases = {
            "Diabetes": ["Frequent urination", "Increased thirst", "Unexplained weight loss"],
            "Hypertension": ["Headache", "Blurred vision", "Shortness of breath"],
            "Flu": ["Fever", "Cough", "Body aches"],
        }
        self.risk_factors = {
            "Smoking": ["Lung Cancer", "Heart Disease"],
            "Obesity": ["Diabetes", "Hypertension"],
            "High Cholesterol": ["Heart Disease", "Stroke"],
        }
        self.treatments = {
            "Diabetes": {"Insulin therapy": 500, "Diet control": 200, "Exercise": 50},
            "Hypertension": {"Medication": 300, "Low-sodium diet": 150, "Exercise": 50},
            "Flu": {"Rest": 0, "Hydration": 20, "Antiviral drugs": 100},
        }

    def identify_risk_factors(self, patient_risks):
        detected_risks = {}
        for risk, diseases in self.risk_factors.items():
            if risk in patient_risks:
                detected_risks[risk] = diseases
        return detected_risks

    def diagnose_disease(self, symptoms):
        probabilities = {}
        for disease, disease_symptoms in self.diseases.items():
            match_count = sum(1 for symptom in symptoms if symptom in disease_symptoms)
            probability = match_count / len(disease_symptoms)  # Simple probability calculation
            probabilities[disease] = probability

        diagnosed_disease = max(probabilities, key=probabilities.get)  # Select disease with highest probability
        return diagnosed_disease, probabilities

    def suggest_treatment(self, disease):
        return self.treatments.get(disease, {"Consult a specialist": 0})

    def plot_risk_factors(self, detected_risks):
        risks = list(detected_risks.keys())
        num_diseases = [len(detected_risks[risk]) for risk in risks]

        plt.figure(figsize=(8, 5))
        plt.bar(risks, num_diseases, color='skyblue')
        plt.xlabel("Risk Factors")
        plt.ylabel("Number of Related Diseases")
        plt.title("Identified Risk Factors and Associated Diseases")
        plt.show()

    def plot_diagnosis_probability(self, probabilities):
        labels = probabilities.keys()
        values = probabilities.values()

        plt.figure(figsize=(7, 7))
        plt.pie(values, labels=labels, autopct="%1.1f%%", startangle=140, colors=['lightcoral', 'lightblue', 'lightgreen'])
        plt.title("Diagnosis Probability Distribution")
        plt.show()

    def plot_treatment_costs(self, treatments):
        treatment_names = list(treatments.keys())
        costs = list(treatments.values())

        plt.figure(figsize=(8, 5))
        plt.bar(treatment_names, costs, color='lightgreen')
        plt.xlabel("Treatment Options")
        plt.ylabel("Cost ($)")
        plt.title("Treatment Cost Comparison")
        plt.show()

# Example usage
if __name__ == "__main__":
    cognitive_healthcare = CognitiveComputingHealthcare()

    patient_symptoms = ["Fever", "Cough"]
    patient_risks = ["Smoking", "Obesity"]

    # Identifying risk factors
    risk_factors = cognitive_healthcare.identify_risk_factors(patient_risks)
    print("Identified Risk Factors and Possible Diseases:", risk_factors)
    cognitive_healthcare.plot_risk_factors(risk_factors)

    # Diagnosing disease
    diagnosed_disease, probabilities = cognitive_healthcare.diagnose_disease(patient_symptoms)
    print("Diagnosed Disease:", diagnosed_disease)
    cognitive_healthcare.plot_diagnosis_probability(probabilities)

    # Providing timely care
    treatment_plan = cognitive_healthcare.suggest_treatment(diagnosed_disease)
    print("Suggested Treatment and Costs:", treatment_plan)
    cognitive_healthcare.plot_treatment_costs(treatment_plan)
```
