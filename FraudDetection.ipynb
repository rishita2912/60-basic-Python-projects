```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

# Load or simulate the dataset
# For demonstration purposes, let's simulate a dataset with customer and transaction features
np.random.seed(42)
data = {
    'transaction_amount': np.random.uniform(10, 1000, 1000),
    'transaction_time': np.random.uniform(0, 24, 1000),
    'customer_age': np.random.randint(18, 70, 1000),
    'customer_tenure': np.random.uniform(0, 10, 1000),
    'is_fraud': np.random.choice([0, 1], size=1000, p=[0.95, 0.05])  # 5% fraudulent
}

df = pd.DataFrame(data)

# Splitting data into features (X) and target (y)
X = df.drop('is_fraud', axis=1)
y = df['is_fraud']

# Splitting into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Standardize the feature data
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Logistic Regression Model for Fraud Detection
model = LogisticRegression(random_state=42)
model.fit(X_train_scaled, y_train)

# Make predictions
y_pred = model.predict(X_test_scaled)

# Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.2f}")
print("Classification Report:\n", classification_report(y_test, y_pred))

# Simulate real-time fraud detection
def detect_fraud(transaction):
    """Detect if a transaction is fraudulent."""
    transaction_scaled = scaler.transform([transaction])
    prediction = model.predict(transaction_scaled)
    return "Fraudulent" if prediction[0] == 1 else "Legitimate"

# Example real-time transaction
new_transaction = [500, 14, 35, 5]  # [amount, time, age, tenure]
result = detect_fraud(new_transaction)
print(f"New Transaction Status: {result}")
```
